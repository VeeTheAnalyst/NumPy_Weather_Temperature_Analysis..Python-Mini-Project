# Python Lists vs NumPy Arrays
Scenario:
You are working with a small retail business.
The business recorded the number of products sold by two sales representatives over three days.
sales_timothy = [10, 15, 20]
sales_brenda = [5, 10, 15]
The manager wants to know the combined sales for each day.
The expected result is:
[15, 25, 35]


TASK 1: Try With Python Lists

```python
#creating variables
sales_timothy = [10, 15, 20]
sales_brenda = [5, 10, 15]

print(sales_timothy + sales_brenda)
```

```python
''Python concatenated the two lists instead of actually adding them
up to get our expected result.
This happened because Python recognizes the '+' operator under
Lists to be for concatenation and not for element_wise maths. 
```

TASK 2: Convert To Numpy Arrays

```python
sales_timothy_array = np.array(sales_timothy)
sales_brenda_array = np.array(sales_brenda)

print(sales_timothy_array + sales_brenda_array)
```

TASK 3: Compare Both Results

```python
#   Approach       ||     Result
#                  ||
#  Python Lists    ||    [10, 15, 20, 5, 10, 15]
#                  ||
#  NumPy arrays    ||    [15, 25, 35]
#
#
#NumPy is useful here because it allows us to express element_wise 
#mathematical operations directly and faster without having to
#manually write a loop for basic numerical operations.'''
```

# WEATHER TEMPERATURE ANALYSIS WITH NUMPY

PROJECT SCENARIO

Imagine you have been given a small dataset from a weather-monitoring station.
The station recorded daily temperatures for one week.
Your job is to use NumPy to turn the raw numerical data into something that can be analyzed.
Here is your dataset:
temperatures = [27.5, 29.0, 30.5, 28.0, 31.0, 32.5, 30.0]
You have been asked to perform some basic numerical analysis.



PART 1: CREATE THE DATASET

```python
#to get numpy running
import numpy as np

# to create a python list
temperatures = [27.5, 29.0, 30.5, 28.0, 31.0, 32.5, 30.0]

# to convert the list to NumPy array
temperature_array = np.array(temperatures)

#to display result
print(temperature_array)
```

PART 2: UNDERSTAND THE ARRAY

```python
#Question 1
#Is the temperature array 1D or 2D?
#Answer : A 1D array
#
#Question 2
#Would you describe it as a vector or matrix?
#Answer : A Vector

#Question 3
#What is its data type?
#Answer : A float64
print(temperature_array.dtype)

#Question 4
#How many temperature observations are present?
#Answer : 7 observations
print(len(temperature_array))
```

PART 3: PERFORM NUMERICAL OPERATIONS

The weather team wants to compare temperatures.
Create another NumPy array representing the target temperature for each day:
target = [28.0, 28.0, 29.0, 28.0, 30.0, 30.0, 29.0]
Convert it to a NumPy array.

Now calculate:

— Temperature Difference

— Temperature Increase by 2°C


```python
#creating a Numpy array
target = np.array([28.0, 28.0, 29.0, 28.0, 30.0, 30.0, 29.0])

#TASK A - To calculate the temperature difference
temp_difference = temperature_array - target

#TASK B - To calaculate the temperature increase by 2°C
temp_increase = temperature_array + 2

#to display result
print("Temperature Difference:", temp_difference)
print("Temperature Increase 2°C:", temp_increase)
```

# MINI PROJECT CHALLENGE

Imagine the weather station has three locations:
lagos = [27.5, 29.0, 30.5, 28.0, 31.0]
abuja = [25.0, 28.0, 29.5, 30.0, 32.0]
ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]


Your task is to create NumPy arrays for all three locations.
Then answer:
Challenge 1
Which location recorded the highest temperature on each day?


Challenge 2
Create a calculation that increases every recorded temperature by 1°C.


Challenge 3
Create a calculation that compares Lagos temperatures with Abuja temperatures.


Challenge 4
Explain why NumPy becomes increasingly useful as the amount of numerical data grows.


```python
# creating an array
lagos = np.array([27.5, 29.0, 30.5, 28.0, 31.0])
abuja =np.array([25.0, 28.0, 29.5, 30.0, 32.0])
ibadan = np.array([26.5, 28.5, 29.0, 27.5, 30.5])

#=====================================================
#CHALLENGE 1: To get the location with highest record
#=====================================================

locations = np.array([lagos, abuja, ibadan]) #combines it to a 2D array

# to find the position of the highest temp for each day
highest_location = np.argmax(locations,axis = 0)

#to create names for the locations
location_names = np.array(["Lagos", "Abuja", "Ibadan"])

#to loop through
for day, index in enumerate (highest_location, start = 1):
    print("Day", day, ":", location_names[index])
          
```

```python
#=======================================================
#CHALLENGE 2: To increase every recorded temperature by 1°C
#========================================================

lagos_increase = lagos + 1
abuja_increase = abuja + 1
ibadan_increase = ibadan + 1

print(f"""
Lagos : {lagos_increase}
Abuja : {abuja_increase}
Ibadan : {ibadan_increase}""")
```

```python
#=======================================================
#CHALLENGE 3: To compare Lagos temperature with Abuja's
#=======================================================

#subtracting Abuja temp from Lagos temp
#Negative values shows Abuja was warmer
#Positive value shows Lagos was warmer

temperature_difference = lagos - abuja

# to display result
print("Lagos vs Abuja :", temperature_difference)
```

```python
#=========================
#CHALLENGE 4:
#=========================
'''NumPy becomes increasingly useful as the amount of numerical data
grows because as data grows, writing manual loops becomes slow and 
prone to error. NumPy allows many values to be stored and processed
efficiently as arrays. Numerical operations can be performed across
the entire dataset without manually calculating each value.'''
```
